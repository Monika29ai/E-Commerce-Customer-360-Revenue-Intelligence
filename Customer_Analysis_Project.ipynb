```python
# Which products/categories generate the most revenue and profit?

# 1. Customer Distribution by Preferred Category

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='preferred_category_1')

plt.title('Customer Distribution by Preferred Category')
plt.xlabel('Preferred Category')
plt.ylabel('Number of Customers')

plt.xticks(rotation=30, ha='right')
plt.tick_params(axis='x', pad=0)

plt.tight_layout()
plt.show()


# 2. Total Purchases by Category

category_purchases = (
    df.groupby('preferred_category_1')['total_purchases']
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_purchases,
    x='total_purchases',
    y='preferred_category_1'
)

plt.title('Total Purchases by Category')
plt.xlabel('Total Purchases')
plt.ylabel('Preferred Category')

plt.tight_layout()
plt.show()


# 3. Total Spending by Category

category_spending = (
    df.groupby('preferred_category_1')['total_spent_usd']
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_spending,
    x='total_spent_usd',
    y='preferred_category_1'
)

plt.title('Total Customer Spending by Category')
plt.xlabel('Total Spent (USD)')
plt.ylabel('Preferred Category')

plt.tight_layout()
plt.show()


# 4. Total Profit by Category

category_profit = (
    df.groupby('preferred_category_1')['customer_profitability_usd']
    .sum()
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_profit,
    x='customer_profitability_usd',
    y='preferred_category_1'
)

plt.title('Total Customer Profit by Category')
plt.xlabel('Customer Profitability (USD)')
plt.ylabel('Preferred Category')

plt.tight_layout()
plt.show()

```
